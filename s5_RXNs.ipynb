```python
from rdkit import Chem
from rdkit.Chem import AllChem
```

# Sustitución nucleofílica

```python
rxn = AllChem.ReactionFromSmarts("[C:1]Br.[OH-:2]>>[C:1]O.[Br-:2]")
```

```python
react1 = Chem.MolFromSmiles("CBr")
react2 = Chem.MolFromSmiles("[OH-]")
```

```python
productos = rxn.RunReactants((react1, react2))
```

```python
prod_smiles1 = Chem.MolToSmiles(productos[0][0])
prod_smiles1
```

```python
prod_smiles2 = Chem.MolToSmiles(productos[0][1])
prod_smiles2
```
